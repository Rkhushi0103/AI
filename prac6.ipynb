Probability Reasoning (Bayesian Networks & Inference)

1] Using a simple example (Burglary–Alarm), create a Bayesian Network using pgmpy.

```python
# Bayesian Network using pgmpy
# Example: Burglary-Alarm Network
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm"),
    ("Alarm", "JohnCalls"),
    ("Alarm", "MaryCalls")
])

# Step 2: Define Conditional Probability Tables
# P(Burglary)
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={
        "Burglary": ["False", "True"]
    }
)

# P(Earthquake)
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={
       "Earthquake": ["False", "True"]
    }
)
# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],  # Alarm = False
        [0.001, 0.29, 0.94, 0.95]   # Alarm = True
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)

# P(JohnCalls | Alarm)
cpd_john = TabularCPD(
    variable="JohnCalls",
    variable_card=2,
    values=[
        [0.95, 0.10],  # JohnCalls = False
        [0.05, 0.90]   # JohnCalls = True
    ],

    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "JohnCalls": ["False", "True"]
        "Alarm": ["False", "True"]
    }
)

# P(MaryCalls | Alarm)
cpd_mary = TabularCPD(
   variable="MaryCalls",
    variable_card=2,
    values=[
        [0.99, 0.30],  # MaryCalls = False

        [0.01, 0.70]   # MaryCalls = True
    ],

    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "MaryCalls": ["False", "True"],
       "Alarm": ["False", "True"]
    }
)

# Step 3: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm,
    cpd_john,
    cpd_mary
)

# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())
 
# Step 5: Perform inference
inference = VariableElimination(model)

# Query: Probability of Burglary when John and Mary both call
result = inference.query(

    variables=["Burglary"],

    evidence={

        "JohnCalls": "True",

        "MaryCalls": "True"

    }
)
print("\nProbability of Burglary given JohnCalls=True and MaryCalls=True:")

print(result)
print("KHUSHI T032")
```

Example: Student Performance Bayesian Network

```python
# Bayesian Network using pgmpy

# Example: Student Exam Performance

from pgmpy.models import DiscreteBayesianNetwork

from pgmpy.factors.discrete import TabularCPD

from pgmpy.inference import VariableElimination


# Step 1: Create Bayesian Network structure

model = DiscreteBayesianNetwork([

    ("Study", "Pass"),

    ("Attendance", "Pass"),

    ("Pass", "Result")

])


# Step 2: Define Conditional Probability Tables

# P(Study)

cpd_study = TabularCPD(

    variable="Study",

    variable_card=2,

    values=[[0.6], [0.4]],

    state_names={

        "Study": ["Low", "High"]

    }

)


# P(Attendance)

cpd_attendance = TabularCPD(

    variable="Attendance",

    variable_card=2,

    values=[[0.3], [0.7]],

    state_names={

        "Attendance": ["Low", "High"]

    }

)


# P(Pass | Study, Attendance)

cpd_pass = TabularCPD(

    variable="Pass",

    variable_card=2,

    values=[

        [0.90, 0.60, 0.50, 0.10],  # Pass = No

        [0.10, 0.40, 0.50, 0.90]   # Pass = Yes

    ],

    evidence=["Study", "Attendance"],

    evidence_card=[2, 2],

    state_names={

        "Pass": ["No", "Yes"],

        "Study": ["Low", "High"],

        "Attendance": ["Low", "High"]

    }

)


# P(Result | Pass)

cpd_result = TabularCPD(

    variable="Result",

    variable_card=2,

    values=[

        [0.80, 0.20],  # Result = Fail

        [0.20, 0.80]   # Result = Excellent

    ],

    evidence=["Pass"],

    evidence_card=[2],

    state_names={

        "Result": ["Fail", "Excellent"],

        "Pass": ["No", "Yes"]

    }

)


# Step 3: Add CPDs to model

model.add_cpds(

    cpd_study,

    cpd_attendance,

    cpd_pass,

    cpd_result

)


# Step 4: Check model correctness

print("Is Bayesian Network valid?", model.check_model())


# Step 5: Perform inference

inference = VariableElimination(model)


# Query: Probability of Pass when Study and Attendance are High

result = inference.query(

    variables=["Pass"],

    evidence={

        "Study": "High",

        "Attendance": "High"

    }

)


print("\nProbability of Pass given Study=High and Attendance=High:")

print(result)

print("KHUSHI T032")
```

2. Write a program to compute Posterior Probability using BayesianNetworks(Burglar|Alarm = True)

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Create Bayesian Network
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm")
])

# Probability of Burglary
cpd_burglary = TabularCPD(
    "Burglary", 2, [[0.999], [0.001]],
    state_names={"Burglary": ["False", "True"]}
)

# Probability of Earthquake
cpd_earthquake = TabularCPD(
    "Earthquake", 2, [[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)

# Probability of Alarm
cpd_alarm = TabularCPD(
    "Alarm", 2,
    [[0.999, 0.71, 0.06, 0.05],
     [0.001, 0.29, 0.94, 0.95]],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)

# Add CPDs
model.add_cpds(cpd_burglary, cpd_earthquake, cpd_alarm)

print("Is Bayesian Network valid?", model.check_model())

# Inference
inference = VariableElimination(model)

result = inference.query(
    variables=["Burglary"],
    evidence={"Alarm": "True"}
)

print("\nPosterior Probability:")
print("P(Burglary | Alarm = True)")
print(result)
print("KHUSHI T032")
```

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Create Bayesian Network
model = DiscreteBayesianNetwork([
    ("Study", "Pass"),
    ("Attendance", "Pass")
])

# Probability of Study
cpd_study = TabularCPD(
    "Study", 2,
    [[0.6], [0.4]],
    state_names={"Study": ["Low", "High"]}
)

# Probability of Attendance
cpd_attendance = TabularCPD(
    "Attendance", 2,
    [[0.3], [0.7]],
    state_names={"Attendance": ["Low", "High"]}
)

# Probability of Pass
cpd_pass = TabularCPD(
    "Pass", 2,
    [[0.90, 0.60, 0.50, 0.10],
     [0.10, 0.40, 0.50, 0.90]],
    evidence=["Study", "Attendance"],
    evidence_card=[2, 2],
    state_names={
        "Pass": ["No", "Yes"],
        "Study": ["Low", "High"],
        "Attendance": ["Low", "High"]
    }
)

# Add CPDs
model.add_cpds(cpd_study, cpd_attendance, cpd_pass)

# Check model
print("Is Bayesian Network valid?", model.check_model())

# Perform inference
inference = VariableElimination(model)

result = inference.query(
    variables=["Pass"],
    evidence={
        "Study": "High",
        "Attendance": "High"
    }
)

print("\nPosterior Probability:")
print("P(Pass | Study = High, Attendance = High)")
print(result)
print("KHUSHI T032")
```
